```python

servers = [
    {
        "server_id": "server-01",
        "cpu_usage": 95,
        "memory_usage": 82,
        "status": "running"
    },
    {
        "server_id": "server-02",
        "cpu_usage": 35,
        "memory_usage": 48,
        "status": "running"
    },
    {
        "server_id": "server-03",
        "cpu_usage": 60,
        "memory_usage": 96,
        "status": "running"
    },
    {
        "server_id": "server-04",
        "cpu_usage": 20,
        "memory_usage": 30,
        "status": "stopped"
    }
]

for server in servers:
    print(server)
```

```python

def analyze_server(server):
    issues = []

    if server["status"] != "running":
        issues.append("Server is not running")

    if server["cpu_usage"] > 80:
        issues.append("High CPU usage")

    if server["memory_usage"] > 90:
        issues.append("High memory usage")

    if not issues:
        return {
            "server_id": server["server_id"],
            "status": "Healthy",
            "issues": []
        }

    return {
        "server_id": server["server_id"],
        "status": "Incident",
        "issues": issues
    }


for server in servers:
    result = analyze_server(server)
    print(result)
```

```python

# Test 1: High CPU should be detected
test_server = {
    "server_id": "test-01",
    "cpu_usage": 95,
    "memory_usage": 50,
    "status": "running"
}

result = analyze_server(test_server)

assert result["status"] == "Incident"
assert "High CPU usage" in result["issues"]


# Test 2: Healthy server should have no issues
healthy_server = {
    "server_id": "test-02",
    "cpu_usage": 30,
    "memory_usage": 40,
    "status": "running"
}

result = analyze_server(healthy_server)

assert result["status"] == "Healthy"
assert result["issues"] == []


# Test 3: Stopped server should be detected
stopped_server = {
    "server_id": "test-03",
    "cpu_usage": 10,
    "memory_usage": 20,
    "status": "stopped"
}

result = analyze_server(stopped_server)

assert result["status"] == "Incident"
assert "Server is not running" in result["issues"]

print("All tests passed!")
```

```python

import pandas as pd

results = [analyze_server(server) for server in servers]

df = pd.DataFrame([
    {
        "server_id": result["server_id"],
        "status": result["status"],
        "issues": ", ".join(result["issues"]) or "None"
    }
    for result in results
])

df
```
